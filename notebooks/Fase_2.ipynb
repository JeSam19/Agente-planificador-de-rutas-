```python
# Práctica 1: BÚSQUEDA (Fase 2: Búsqueda Informada)

## Planteamiento

En esta fase se aborda la búsqueda de caminos mínimos incorporando conocimiento geográfico mediante **heurísticas admisibles**. El objetivo es demostrar empíricamente que una heurística admisible reduce la cantidad de nodos expandidos respecto a la búsqueda a ciegas sin perder la optimalidad del camino $A^*$, y comparar su desempeño contra **Greedy Best-First Search**.

Se evalúan tres heurísticas admisibles:
1. **$h_1$ - Distancia Euclidiana:** Proyección cartesiana en plano de dos dimensiones.
2. **$h_2$ - Distancia Haversine:** Distancia sobre la superficie esférica terrestre.
3. **$h_3$ - Heurística Personalizada:** Combinación de Haversine con penalización de intersecciones.
```

```python
import sys
from pathlib import Path

# Configurar la ruta raíz del proyecto
RAIZ = Path.cwd().resolve().parent if Path.cwd().name == "notebooks" else Path.cwd().resolve()
if str(RAIZ) not in sys.path:
    sys.path.append(str(RAIZ))

import pandas as pd
import folium
from src.utils import cargar_grafo, gen_nodo_entrega
from src.fase1 import accesible
from src.fase2 import (
    h1_euclidiana,
    h2_haversine,
    h3_personalizada,
    a_star,
    greedy_best_first
)

print("✓ Módulos importados correctamente.")
```

```python
## 1. Carga del Grafo Urbano y Selección de Instancias
Cargamos la red vial de las alcaldías descargadas previamente y seleccionamos un par de nodos origen-destino alcanzables.
```

```python
ruta_grafo = RAIZ / "data" / "cdmx_norte_centro.graphml"
print(f"Buscando archivo en: {ruta_grafo}")

G = cargar_grafo(str(ruta_grafo))

# Selección de origen y destino válidos
nodos = list(G.nodes())
origen = nodos[10]
destino = nodos[800]

print(f"Nodo Origen: {origen}")
print(f"Nodo Destino: {destino}")
```

```python
## 2. Ejecución de $A^*$ y Greedy Best-First Search
```

```python
print("Ejecutando A* con h1 (Euclidiana)...")
res_a1 = a_star(G, origen, destino, h1_euclidiana)

print("Ejecutando A* con h2 (Haversine)...")
res_a2 = a_star(G, origen, destino, h2_haversine)

print("Ejecutando A* con h3 (Personalizada)...")
res_a3 = a_star(G, origen, destino, h3_personalizada)

print("Ejecutando Greedy Best-First Search...")
res_g = greedy_best_first(G, origen, destino, h2_haversine)

print("✓ Algoritmos ejecutados exitosamente.")
```

```python
## 3. Visualización de Rutas en Mapa Interactivo
```

```python
lat_orig, lon_orig = G.nodes[origen]['y'], G.nodes[origen]['x']

m = folium.Map(
    location=[lat_orig, lon_orig],
    zoom_start=14,
    tiles='https://{s}.tile.opentopomap.org/{z}/{x}/{y}.png',
    attr='Map data: &copy; OpenStreetMap | Style: OpenTopoMap'
)

# Ruta Óptima de A* (Azul)
coords_a = [(G.nodes[n]['y'], G.nodes[n]['x']) for n in res_a2['camino']]
folium.PolyLine(coords_a, color="blue", weight=5, opacity=0.8, tooltip="Ruta A* (Óptima)").add_to(m)

# Ruta Greedy (Roja)
coords_g = [(G.nodes[n]['y'], G.nodes[n]['x']) for n in res_g['camino']]
folium.PolyLine(coords_g, color="red", weight=3, opacity=0.8, tooltip="Ruta Greedy (No Óptima)").add_to(m)

m.save(str(RAIZ / "visualizacion_rutas_informadas.html"))
m
```

```python
## 4. Tabla Comparativa de Desempeño y Exportación
```

```python
df_resultados = pd.DataFrame([
    {
        "Algoritmo": "A* (Euclidiana h1)",
        "Nodos Expandidos": res_a1['nodos_expandidos'],
        "Costo Total (m)": round(res_a1['costo'], 2),
        "Max Frontera": res_a1['max_frontera'],
        "Tiempo (ms)": f"{res_a1['tiempo_ms']:.2f}"
    },
    {
        "Algoritmo": "A* (Haversine h2)",
        "Nodos Expandidos": res_a2['nodos_expandidos'],
        "Costo Total (m)": round(res_a2['costo'], 2),
        "Max Frontera": res_a2['max_frontera'],
        "Tiempo (ms)": f"{res_a2['tiempo_ms']:.2f}"
    },
    {
        "Algoritmo": "A* (Personalizada h3)",
        "Nodos Expandidos": res_a3['nodos_expandidos'],
        "Costo Total (m)": round(res_a3['costo'], 2),
        "Max Frontera": res_a3['max_frontera'],
        "Tiempo (ms)": f"{res_a3['tiempo_ms']:.2f}"
    },
    {
        "Algoritmo": "Greedy Best-First",
        "Nodos Expandidos": res_g['nodos_expandidos'],
        "Costo Total (m)": round(res_g['costo'], 2),
        "Max Frontera": res_g['max_frontera'],
        "Tiempo (ms)": f"{res_g['tiempo_ms']:.2f}"
    }
])

df_resultados.to_csv(str(RAIZ / "reporte_busqueda_informada.csv"), index=False)
display(df_resultados)
```
